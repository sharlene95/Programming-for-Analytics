```python
#Task 3 - Filtering and Selection
#Filter to find warriors with Battles_Fought > 7 and Avg_Damage > 85
df[(df['Battles_Fought'] > 7) & (df['Avg_Damage'] > 85)]
```

```python
#Create a new DataFrame of these elite fighters and display it
elite_fighters = df[(df['Battles_Fought'] > 7) & (df['Avg_Damage'] > 85)]

elite_fighters
```

```python
#Sort the result by Avg_Damage in descending order

elite_fighters.sort_values(by= 'Avg_Damage', ascending=False)
```

```python
#Task 4 - Create a Calculated Column
#Add a new column called Battle_Efficiency

df['Battle_Efficiency'] = (df['Avg_Damage'] * df['Battles_Fought']) / (df['Injuries_Sustained'] + 1)

df.head()
```

```python
#Task 5 - Grouping and Saving Results
#Group data by Role and calculate the average Battle_Efficiency

bridgeburners_summary = df.groupby('Role')['Battle_Efficiency'].mean().round(2)

bridgeburners_summary
```

```python
#Save the grouped summary to a new file bridgeburners_summary.csv

bridgeburners_summary.to_csv('bridgeburners_summary.csv')
```
