```python
#1. Read 'grades.txt' file and parse each line into a dictionary

grades = {}

with open("grades.txt", "r") as f:
    for line in f:
        name, grade = line.split(",")
        grades[name] = int(grade)

print(grades)
```

```python
#2. Calculate and print average grade

average_grade = sum(grades.values()) / len(grades)

print(f"Average Grade: {average_grade:.2f}")
```

```python
#3. Save a new file 'above_average.csv' containing only the names and grades of students who scored above average

import csv

with open("above_average.csv", "w") as f:
    writer = csv.writer(f)

    writer.writerow(["name", "grade"])

    for name, grade in grades.items():
        if grade > average_grade:
            writer.writerow([name, grade])
```

```python
#4. Save all student data in a JSON file with format

import json

students = []

for name, grade in grades.items():
    students.append({"name": name, "grade": grade})

with open("students.json", "w") as json_file:
    json.dump(students, json_file)

print(students)
```

```python
#5. Wrap all file reading/writing in try/except blocks with appropriate error messages
#Wrap reading grades.txt

grades = {}

try:
    with open("grades.txt", "r") as f:
        for line in f:
            name, grade = line.split(",")
            grades[name] = int(grade)

    print(grades)

except FileNotFoundError:
    print("Error: grades.txt was not found.")
```

```python
#Wrap calculating average grade

if not grades:
    print("Error: No student grades found.")

else:
    average_grade = sum(grades.values()) / len(grades)
    print(f"Average Grade: {average_grade:.2f}")
```

```python
#Wrap writing above_average csv

try:
    with open("above_average.csv", "w") as f:
        writer = csv.writer(f)

        writer.writerow(["name", "grade"])

        for name, grade in grades.items():
            if grade > average_grade:
                writer.writerow([name, grade])

except IOError:
    print("Error: Could not write above_average.csv.")
```

```python
#Wrap writing students.json file

students = []

for name, grade in grades.items():
    students.append({"name": name, "grade": grade})

try:
    with open("students.json", "w") as json_file:
        json.dump(students, json_file)

    print(students)

except IOError:
    print("Error: Could not write students.json.")
```
