```python
#1. Read the 'grades.txt' file and parse each line into a dictionary with keys: name and grade.

file_path = "grades.txt"

students = []

try:
    with open(file_path, "r") as f:
        for line in f:
            name, grade = line.strip().split(",")
            students.append({
                "name": name.strip(),
                "grade": int(grade.strip())
            })

except FileNotFoundError:
    print(f"Error: File '{file_path}' not found.")

except Exception as e:
    print(f"An error occurred: {e}")

students
```

```python
#2. Use the 'https://api.agify.io?name=FIRSTNAME' API to estimate the age of each student. Extract only the first name for the API query. Add the predicted age as a new field.

import requests

base_url = "https://api.agify.io"

for student in students:

    if student.get("age") is None:
        first_name = student["name"].split()[0]

        params = {"name": first_name}
        
        try:
            response = requests.get(base_url, params)
            
            response.raise_for_status()
            
            data = response.json()

            student_age = data.get("age")

            student["age"] = student_age
        
        except (requests.RequestException, ValueError) as e:
            print(f"Error retrieving age for {first_name}: {e}")
            student["age"] = None

students      
```

```python
#3. Calculate the average grade. For each student, add a new field: status = 'Over' if grade >= avg, else 'Under'.

if students:
    total = sum(student["grade"] for student in students)
    average_grade = total / len(students)

    print(f"Average grade: {average_grade:.2f}")

    for student in students:
        if student["grade"] >= average_grade:
            student["status"] = "Over"

        else:
            student["status"] = "Under"

else:
    average_grade = 0

students
```

```python
#4. Save all students with ages and statuses into a JSON file called 'students_enriched.json'.

import json

try:
    with open("students_enriched.json", "w") as f:
        json.dump(students, f, indent=4)

    print("students_enriched.json created successfully.")

except IOError as e:
    print(f"Error saving JSON file: {e}")
```

```python
#5. Write a CSV file called 'over_students.csv' with only students who scored above average. Include name, grade, age.

import csv

try:
    with open("over_students.csv", "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames = ["name", "grade", "age"])
        writer.writeheader()

        for student in students:
            if student["grade"] > average_grade:
                writer.writerow({
                    "name": student["name"],
                    "grade": student["grade"],
                    "age": student["age"]
                })

    print("over_students.csv created successfully.")

except IOError as e:
    print(f"Error writing CSV file: {e}")
```
